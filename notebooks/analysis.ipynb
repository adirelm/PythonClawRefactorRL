# A4 Phase-4 analysis (D7 + heatmap + marginals)

Reads SDK `Ablation` object + embeds figures. **No parallel implementation** — only the SDK.

```python
from pathlib import Path
import sys
REPO = Path('.').resolve()
if str(REPO) not in sys.path:
    sys.path.insert(0, str(REPO))
from src.sdk import Ablation, CellResult, run_ablation
ab = run_ablation('compact', output_dir=Path('results/ablations'))
print(f"cells: {len(ab.cells)}  n_ok_total: {ab.n_ok_total}  wall_clock_s: {ab.total_wall_clock_s:.1f}")
```

## Best / worst / baseline cells

```python
# Find best/worst/baseline
cells_full = [c for c in ab.cells if c.n_ok >= 3]
best = max(cells_full, key=lambda c: c.mean_final_reward)
worst = min(cells_full, key=lambda c: c.mean_final_reward)
baseline = next(c for c in ab.cells if (c.alpha, c.beta, c.gamma, c.p_skills) == (1.0, 1.0, 0.5, -5.0))
for tag, c in [('best', best), ('worst', worst), ('baseline', baseline)]:
    print(f"{tag:9s}: a={c.alpha} b={c.beta} g={c.gamma} P={c.p_skills} mean={c.mean_final_reward:+.4f} ci95+/-{c.ci95_halfwidth:.4f} n_ok={c.n_ok}")
```
